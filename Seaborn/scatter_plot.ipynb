```python
# Import required libraries for data visualization and manipulation
# matplotlib.pyplot: For creating plots and customizing figures
# numpy: For numerical operations
# seaborn: For statistical data visualization (built on matplotlib)
# pandas: For data manipulation and analysis
import matplotlib.pyplot as plt 
import numpy as np 
import seaborn as sns 
import pandas as pd
```

```python
# Load the 'tips' dataset from seaborn and display its shape
# This dataset contains information about restaurant tips with 244 rows and 7 columns
df = sns.load_dataset('tips')
df.shape  # Display the dimensions: (number of rows, number of columns)
```

```python
# Display the first few rows of the dataset to understand its structure
# This gives a quick preview of the data columns and values
df.head()
```

```python
# Display detailed information about the dataset including column types and memory usage
# Helps identify data types (int, float, object) and check for missing values per column
df.info()
```

```python
# Check for missing values in each column
# .isna() returns True for missing values, .sum() counts them per column
# Result of 0 means no missing values exist
df.isna().sum()
```

```python
# Create a basic scatter plot showing the relationship between total_bill and tip
# x-axis: total_bill (amount of the restaurant bill)
# y-axis: tip (amount of tip left)
# This simple plot helps visualize the correlation between bill amount and tip amount
sns.scatterplot(x='total_bill', y='tip', data=df)
plt.show()
```

```python
# Create a scatter plot with hue parameter to distinguish by gender
# hue='sex': Different colors for 'Male' and 'Female'
# This allows us to compare tipping patterns between genders
# Added title for better clarity
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=df)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python
# Create a more complex scatter plot with multiple visual dimensions
# hue='sex': Color coded by gender (Male/Female)
# style='smoker': Different marker styles for smokers vs non-smokers
# This reveals patterns based on both gender and smoking status
# Added axis labels for better readability
sns.scatterplot(x='total_bill', y='tip', hue='sex', style='smoker', data=df)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
# Create a scatter plot with four visual dimensions:
# x-axis: total_bill, y-axis: tip, hue='sex': color by gender
# style='smoker': marker style, size='size': party size determines point size
# legend='full': Show complete legend for all variables
# This comprehensive visualization shows relationships with multiple variables
sns.scatterplot(x='total_bill', y='tip', hue='sex', style='smoker', size='size', legend='full', data=df)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
# Enhanced scatter plot with custom color palette
# palette='Set1': Uses a predefined color scheme for better visual appeal
# This version is more readable with better organized code formatting
# Demonstrates how different hue colors can improve data visualization
sns.scatterplot(
    x='total_bill', 
    y='tip',
    hue='sex',
    style='smoker',
    size='size',
    legend='full',
    palette='Set1',  # Custom color palette for better aesthetics
    data=df
)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
# Final comprehensive scatter plot with all customizations
# markers=True: Enable marker shapes (though this parameter is less commonly used)
# This represents the most detailed and polished version of the visualization
# Shows best practices for creating publication-ready data visualizations
# The plot effectively communicates relationships between multiple variables:
# - Total bill vs tip amount
# - Color difference by gender
# - Marker style difference by smoker status  
# - Point size represents party size
sns.scatterplot(
    x='total_bill', 
    y='tip',
    hue='sex',
    style='smoker',
    size='size',
    legend='full',
    palette='Set1',
    markers=True,  # Explicitly enable marker customization
    data=df
)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```
